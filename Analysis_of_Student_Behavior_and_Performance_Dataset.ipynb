# **Some Coding Questions.**

```python
# Create a new column Performance_Category based on Total_Score:
# "High" if Total_Score > 80, "Medium" if 50 ≤ Total_Score ≤ 80, and "Low" if Total_Score < 50.

ds["Performance_Category"]=np.where(ds["Total_Score"]>80,"High",np.where(ds["Total_Score"]>=50,"Medium","Low"))
ds["Performance_Category"]
```

```python
# Create a new column Sleep_Category based on Sleep_Hours_per_Night:
# <6 = "Low", 6-8 = "Normal", >8 = "High".

ds["Sleepy_category"]=np.where(ds["Sleep_Hours_per_Night"]<6,"Low", np.where(ds["Sleep_Hours_per_Night"]<=8,"Normal","High"))
ds["Sleepy_category"]
```

```python
# Combine Participation_Score, Quizzes_Avg, and Assignments_Avg into a new column Engagement_Score using their average.

ds["Engagement_Score"]=ds[['Participation_Score', 'Quizzes_Avg', 'Assignments_Avg']].mean(axis=1)
ds["Engagement_Score"]
```

```python
# Calculate the GPA based on grade column.

grade_to_gpa={"A":4.0,"B":3.0,"C":2.0,"D":1.0,"E":0.0}
ds["GPA"]=ds["Grade"].map(grade_to_gpa)
ds["GPA"]
```

```python
# Detect an at risk student.

ds["At Risk"]=np.where((ds["Final_Score"]<60) & (ds["Sleep_Hours_per_Night"]<5) & (ds["Internet_Access_at_Home"]=="No"),"Yes","No")

ds[ds["At Risk"]=="Yes"]
```

```python
# How does family income affect performance?

perform_inc=ds.groupby("Family_Income_Level")["Total_Score"].mean().sort_values()
print("Performnace by family income:\n", perform_inc)

# Not much difference is seen.
```

# **Numpy Questions**

```python
# Using a NumPy array of student ages, calculate the mean age of all students.

age=np.array(ds["Age"])
mean_age=np.mean(age)
print("The mean age is:", mean_age)
```

```python
# From a NumPy array of attendance percentages, extract all students who have attendance less than 75%.

arr=np.array(ds["Attendance (%)"])
att=arr[arr<75]
print("Students with low attendance are:",att)
```

```python
# Create a Boolean mask to identify students who have a Stress_Level (1-10) greater than 7.

arr=np.array(ds["Stress_Level (1-10)"])
mask=arr>7
print("Students with high stress level:", mask)
print("Stress Values:", arr[mask])
```

```python
# 7. Using NumPy arrays, filter students who: Have Attendance (%) ≥ 80, AND Sleep_Hours_per_Night ≥ 6, AND Stress_Level ≤ 5. Return their indices or count.

arr1=np.array(ds["Attendance (%)"])
arr2=np.array(ds["Sleep_Hours_per_Night"])
arr3=np.array(ds["Stress_Level (1-10)"])

fil=(arr1>=80) & (arr2>=6) & (arr3<=5)
print("Filtered indices:",np.where(fil)[0])
print("Matching students:",fil.sum())
```

```python
# 8. From numerical columns like Midterm_Score, Final_Score, Assignments_Avg, Projects_Score, etc.,
# create a NumPy 2D array and compute the correlation matrix using np.corrcoef.

arr1=np.array(ds["Midterm_Score"])
arr2=np.array(ds["Final_Score"])
arr3=np.array(ds["Assignments_Avg"])
arr4=np.array(ds["Projects_Score"])

data_matrix=np.vstack((arr1, arr2, arr3, arr4))
corr_matrix=np.corrcoef(data_matrix)
print("Correlation Matrix:\n", corr_matrix)
```

```python
# 9. Calculate a custom total score using this formula:
# Custom Score=0.2×Midterm+0.3×Final+0.2×Assignments+0.15×Projects
# Use NumPy's vectorized operations.

arr1=np.array(ds["Midterm_Score"])
arr2=np.array(ds["Final_Score"])
arr3=np.array(ds["Assignments_Avg"])
arr4=np.array(ds["Projects_Score"])

score=0.2*arr1+0.3*arr2+0.2*arr3+0.15*arr4
print("Custom Score:\n", score)
```

```python
# 10. Using the IQR method (Interquartile Range), detect outliers in the Study_Hours_per_Week column.
# Outlier=Q1−1.5×IQR or =Q3+1.5×IQR

arr=np.array(ds["Study_Hours_per_Week"])
q1=np.percentile(arr, 25)
q3=np.percentile(arr, 75)
iqr=q3-q1

lower_bound=q1-1.5*iqr
upper_bound=q3-1.5*iqr

outliers=arr[(arr < lower_bound) | (arr > upper_bound)]
print("Outliers in Study Hours:", outliers)
```

# **Data Transformation**

```python
# Normalize the Midterm_Score array using Z-score normalization.

arr=np.array(ds["Midterm_Score"])
mean=np.mean(arr)
std=np.std(arr)
norm_arr=(arr-mean)/std
print("Normalized Midterm Scores:", norm_arr)
```

```python
# Which numerical features have the highest correlation with Total_Score?

num_col=ds.select_dtypes(include=np.number).columns

correlation=ds[num_col].corr()["Total_Score"].sort_values(ascending=False)
print("Correlation with Total Score:\n", correlation)
```

```python
# How does attendance correlate to the Total scores?

correlation = ds['Attendance (%)'].corr(ds['Total_Score'])
print("Correlation between Attendance and Total_Score:",correlation)
```
